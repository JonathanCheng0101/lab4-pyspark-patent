````python
rddCitations = sc.textFile("cite75_99.txt.gz")
rddPatents = sc.textFile("apat63_99.txt.gz")
````

In other words, they are a single string with multiple CSV's. You will need to convert these to (K,V) pairs, probably convert the keys to `int` and so on. You'll need to `filter` out the header string as well since there's no easy way to extract all the lines except the first.

````python
# Remove header

headerC = rddCitations.first()
headerP = rddPatents.first()

citations = rddCitations.filter(lambda x: x != headerC)
patents   = rddPatents.filter(lambda x: x != headerP)


````

````python
# Split the strings into (citing, cited) format

def parse_citation(line):
    parts = line.split(",")
    return (int(parts[0]), int(parts[1]))

citKV = citations.map(parse_citation)
# (citing, cited)

````

````python
# Parse patents: --> (patent_id, (country, state))

def parse_patent(line):
    parts = line.split(",")
    patent_id = int(parts[0])
    country = parts[4].strip('"')
    state = parts[5].strip('"')
    return (patent_id, (country, state))

patKV = (
    patents
    .map(parse_patent)
    .filter(lambda x: x[1][0] == "US" and x[1][1] != "")
    .cache()
)

# x[1][1] !="" --> has to have state name
# cache the (patent_id, ("US", "CA")) result for future use

````

````python
cited_join = citKV.map(lambda x: (x[1], x[0])) \
    .join(patKV)
# (cited, (citing, (country, cited_state)))

````

````python
# Join 2 add state of Citing, the df turns to (cited, (citing, (country, cited_state)))
both_states = cited_join.map(
    lambda x: (x[1][0], (x[0], x[1][1][1]))  # (citing, (cited, cited_state))
).join(patKV)
# (citing, ((cited, cited_state), (country, citing_state)))

both_states = both_states.cache()

````

````python
#filter self-state

self_state = both_states.filter(
    lambda x: x[1][0][1] == x[1][1][1]
)
# cited_state == citing_state

````

````python
# count patents of citing

self_counts = self_state.map(
    lambda x: (x[0], 1)
).reduceByKey(operator.add)

````

````python
# get top 10 (based on count desc first, then patent ascenden
top10 = self_counts.sortBy(lambda x: (-x[1], x[0])).take(10)

for t in top10:
    print(t)
````
